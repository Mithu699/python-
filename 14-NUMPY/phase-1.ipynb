### creating array from scratch


```python
zeros = np.zeros((3,4))
print("zeros array : \n ", zeros)

ones = np.ones((3,4))
print("ones array: \n ", ones)

full = np.full((2,2), 7) # special constant ka milega 
print("full array: \n", full)

random = np.random.random((2,4))
print("random array: \n", random)

sequence = np.arange(0 , 10, 2)
print("sequence array : \n", sequence)
```

### vector , matrix and tensor 

```python
import numpy as np
# tensor me zyada dimension hota hai isiliye complex lagta hai 
vector = np.array([1,2,3])
print("vector: ", vector)

matrix = np.array([[1,2,3],
                  [4,5,6]])
print("matrix: ", matrix)

tensor = np.array([[[1,2], [3,4]],
                   [[5,6], [7,8]]])

print("tensor array: \n", tensor)
```

### array properties


```python
import numpy as np

arr = np.array([[1,2,3],
                [4,5,6]])
print("Shape", arr.shape)
print("Dimension", arr.ndim)
print("Size", arr.size)
print("Dtype", arr.dtype)
```

### ARRAY reshaping 


```python
arr = np.arange(12)
print("Original array", arr)

reshaped = arr.reshape((3,4))
print("\n Reshaped array" , reshaped)

flattend = reshaped.flatten()
print("\n Flattened array", flattend)

# ravel = return view , instead of copy

raveled = reshaped.ravel()
print("\n raveled array ", raveled)

#transpose

transpose = reshaped.T
print("\n Transposed array", transpose)
```
